# Análise exploratória — Histórico de vendas (Saltim)

Notebook para inspecionar o dataset sintético `vendas.csv` (2023-01-01 a 2026-05-19).

**Fontes:**
- `vendas.csv` — transações (id, date_time, recipe_id, quantity, unit_price)
- `receitas.csv` — nomes, tipo e preços dos produtos finais
- `resumo_diario_vendas.csv` e `resumo_mensal_vendas.csv` — indicadores agregados
- `feriados_recife.csv` — calendário de feriados de Recife
- `produtos_indisponiveis.csv` — períodos de indisponibilidade de produtos

## 1. Setup e carga dos dados

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

_HERE = Path.cwd().resolve()
_CANDIDATES = [
    _HERE,
    *(_HERE.parents),
    *[parent / "data" for parent in [_HERE, *_HERE.parents]],
]
DATA_DIR = next((p for p in _CANDIDATES if (p / "vendas.csv").exists()), None)
if DATA_DIR is None:
    raise FileNotFoundError(
        "Não foi possível localizar 'vendas.csv'. "
        "Confirme que os CSVs estão na pasta 'data/' do repositório."
    )

sns.set_theme(style="whitegrid", palette="deep")
plt.rcParams.update({
    "figure.figsize": (12, 5),
    "axes.titlesize": 13,
    "axes.labelsize": 11,
})

MESES_PT = {
    1: "Jan", 2: "Fev", 3: "Mar", 4: "Abr", 5: "Mai", 6: "Jun",
    7: "Jul", 8: "Ago", 9: "Set", 10: "Out", 11: "Nov", 12: "Dez",
}
DIAS_PT = ["Seg", "Ter", "Qua", "Qui", "Sex", "Sáb", "Dom"]

vendas = pd.read_csv(DATA_DIR / "vendas.csv", parse_dates=["date_time"])
receitas = pd.read_csv(DATA_DIR / "receitas.csv")
resumo_diario = pd.read_csv(DATA_DIR / "resumo_diario_vendas.csv", parse_dates=["date"])
resumo_mensal_path = DATA_DIR / "resumo_mensal_vendas.csv"
resumo_mensal = pd.read_csv(resumo_mensal_path) if resumo_mensal_path.exists() else None
feriados = pd.read_csv(DATA_DIR / "feriados_recife.csv", parse_dates=["data"])
indisp = pd.read_csv(DATA_DIR / "produtos_indisponiveis.csv", parse_dates=["data_inicio", "data_fim"])

produtos = receitas[receitas["type"] == "PRODUTO_FINAL"][["id", "name", "sale_price"]].copy()
produtos = produtos.rename(columns={"id": "recipe_id", "name": "produto", "sale_price": "preco"})

# Categorização leve do produto (mesma lógica do gerador de vendas)
def categorizar(nome: str) -> str:
    n = str(nome).upper()
    if "INFANTIL" in n:
        return "Infantil"
    if any(t in n for t in ["SHAKE", "SODA", "BOWL", "AÇAÍ", "ACAI", "GINGER", "GREENTEA", "LEMON GRASS", "FRUTAS VERMELHAS"]):
        return "Bebida fria"
    if any(t in n for t in ["ESPRESSO", "AFFOGATO", "CAPPUCCINO", "LATTE", "MOCHA", "TONIC", "CAFÉ", "CAFE"]):
        return "Bebida quente"
    if any(t in n for t in ["BROWNIE", "BLONDIE", "BOLO", "BRIGADEIRO", "CHEESECAKE", "BANNOFFE", "LEMON BAR", "SORVETE"]):
        return "Sobremesa"
    if any(t in n for t in ["TOAST", "RABANADA", "PANQUECA", "OVERNIGHT", "OATS", "CUSCUZ", "BRIOCHE", "CROISSANT", "GRANOLA"]):
        return "Café da manhã"
    if any(t in n for t in ["SUB", "KATSU", "PF ", "LINGUINE", "CROQUE", "ARROZ", "PARMÊ", "PARME", "EXECUTIVO", "BIFUM", "COXINHA", "TAPIOCA", "KLEBINHO", "RECIFE É UM OVO"]):
        return "Almoço"
    return "Outros"

produtos["categoria"] = produtos["produto"].apply(categorizar)

df = vendas.merge(produtos, on="recipe_id", how="left")
df["data"] = df["date_time"].dt.floor("D")
df["ano"] = df["date_time"].dt.year
df["mes"] = df["date_time"].dt.month
df["dia_semana"] = df["date_time"].dt.weekday
df["hora"] = df["date_time"].dt.hour
df["receita_linha"] = df["quantity"] * df["unit_price"]
df["categoria"] = df["categoria"].fillna("Outros")

print(f"DATA_DIR = {DATA_DIR}")
print(f"Registros: {len(df):,}")
print(f"Período: {df['date_time'].min()} → {df['date_time'].max()}")
print(f"Produtos distintos: {df['recipe_id'].nunique()}")
df.head()
```

## 2. Visão geral (KPIs)

```python
kpis = pd.DataFrame(
    {
        "Métrica": [
            "Total de vendas (linhas)",
            "Unidades vendidas",
            "Dias com movimento",
            "Produtos distintos vendidos",
            "Receita total (R$)",
            "Ticket médio por venda (R$)",
            "Itens por venda (média)",
            "Preço unitário médio (R$)",
            "Maior share de produto",
        ],
        "Valor": [
            f"{len(df):,}",
            f"{int(df['quantity'].sum()):,}",
            f"{df['data'].nunique():,}",
            f"{df['recipe_id'].nunique():,}",
            f"R$ {df['receita_linha'].sum():,.2f}",
            f"R$ {df['receita_linha'].mean():.2f}",
            f"{df['quantity'].mean():.2f}",
            f"R$ {df['unit_price'].mean():.2f}",
            f"{df['recipe_id'].value_counts(normalize=True).iloc[0]:.1%}",
        ],
    }
)

display(kpis)

por_ano = (
    df.groupby("ano")
    .agg(
        vendas=("id", "count"),
        unidades=("quantity", "sum"),
        receita=("receita_linha", "sum"),
        ticket_medio=("receita_linha", "mean"),
    )
    .assign(var_vendas_pct=lambda x: x["vendas"].pct_change() * 100)
)
por_ano.round(2)
```

## 3. Tendência temporal

```python
if resumo_mensal is not None:
    mensal = resumo_mensal.copy()
    mensal["periodo"] = pd.to_datetime(
        mensal["year"].astype(str) + "-" + mensal["month"].astype(str).str.zfill(2) + "-01"
    )
else:
    mensal = (
        df.assign(periodo=df["date_time"].dt.to_period("M").dt.to_timestamp())
        .groupby("periodo", as_index=False)
        .agg(vendas_mes=("id", "count"), receita_total=("receita_linha", "sum"))
    )
    mensal["year"] = mensal["periodo"].dt.year

diario = df.groupby("data", as_index=False).agg(
    vendas=("id", "count"),
    receita=("receita_linha", "sum"),
)
diario["mm7"] = diario["vendas"].rolling(7, min_periods=1).mean()

fig, axes = plt.subplots(2, 1, figsize=(14, 9), sharex=False)

axes[0].plot(mensal["periodo"], mensal["vendas_mes"], marker="o", linewidth=2)
axes[0].set_title("Vendas mensais")
axes[0].set_ylabel("Nº de vendas")
axes[0].grid(alpha=0.3)

ax2 = axes[1]
ax2.plot(diario["data"], diario["vendas"], alpha=0.30, linewidth=0.8, label="Vendas diárias")
ax2.plot(diario["data"], diario["mm7"], color="#C44E52", linewidth=2.0, label="Média móvel 7d")
ax2.set_title("Vendas diárias com média móvel")
ax2.set_ylabel("Nº de vendas")
ax2.grid(alpha=0.3)

ax2b = ax2.twinx()
ax2b.plot(diario["data"], diario["receita"].rolling(7, min_periods=1).mean(),
          color="tab:green", alpha=0.55, label="Receita MM7 (R$)")
ax2b.set_ylabel("Receita diária (R$)")

lines_a, labels_a = ax2.get_legend_handles_labels()
lines_b, labels_b = ax2b.get_legend_handles_labels()
ax2.legend(lines_a + lines_b, labels_a + labels_b, loc="upper right")

plt.tight_layout()
plt.show()
```

## 4. Sazonalidade semanal, mensal e horária

```python
vendas_semana = df.groupby("dia_semana").size().reindex(range(7), fill_value=0)
vendas_semana.index = DIAS_PT

heat_mensal = (
    df.groupby(["ano", "mes"]).size().unstack(fill_value=0).reindex(columns=range(1, 13), fill_value=0)
)
heat_mensal.columns = [MESES_PT[m] for m in heat_mensal.columns]

heat_dow_hora = df.groupby(["dia_semana", "hora"]).size().unstack(fill_value=0)
heat_dow_hora = heat_dow_hora.reindex(index=range(7), columns=range(8, 20), fill_value=0)
heat_dow_hora.index = DIAS_PT
heat_dow_hora.columns = [f"{h:02d}h" for h in range(8, 20)]

fig, axes = plt.subplots(1, 3, figsize=(20, 5))

axes[0].bar(vendas_semana.index, vendas_semana.values, color=sns.color_palette("deep")[0])
axes[0].set_title("Vendas por dia da semana")
axes[0].set_ylabel("Nº de vendas")
for i, v in enumerate(vendas_semana.values):
    axes[0].text(i, v + max(vendas_semana.values) * 0.01, f"{int(v):,}", ha="center", fontsize=9)

sns.heatmap(heat_mensal, cmap="Blues", annot=True, fmt=".0f", ax=axes[1], cbar_kws={"label": "Vendas"})
axes[1].set_title("Vendas mensais (ano x mês)")
axes[1].set_xlabel("")
axes[1].set_ylabel("")

sns.heatmap(heat_dow_hora, cmap="YlOrRd", ax=axes[2], cbar_kws={"label": "Vendas"})
axes[2].set_title("Dia da semana × hora")
axes[2].set_xlabel("Hora")
axes[2].set_ylabel("")

plt.tight_layout()
plt.show()
```

## 5. Distribuição por categoria de produto

```python
cat_counts = df[["recipe_id", "categoria"]].drop_duplicates().groupby("categoria")["recipe_id"].count().sort_values(ascending=True)
cat_receita = df.groupby("categoria", as_index=False)["receita_linha"].sum().sort_values("receita_linha", ascending=True)

cat_month = (
    df.assign(periodo=df["date_time"].dt.to_period("M").dt.to_timestamp())
    .groupby(["periodo", "categoria"], as_index=False)["id"].count()
    .rename(columns={"id": "vendas"})
)
cat_pivot = cat_month.pivot(index="periodo", columns="categoria", values="vendas").fillna(0)
share = cat_pivot.div(cat_pivot.sum(axis=1).replace(0, np.nan), axis=0).fillna(0)

fig, axes = plt.subplots(1, 3, figsize=(20, 6))
axes[0].barh(cat_counts.index, cat_counts.values, color=sns.color_palette("deep")[0])
axes[0].set_title("Nº de produtos por categoria")
axes[0].set_xlabel("Produtos distintos")

axes[1].barh(cat_receita["categoria"], cat_receita["receita_linha"], color=sns.color_palette("deep")[1])
axes[1].set_title("Receita acumulada por categoria (R$)")
axes[1].set_xlabel("Receita (R$)")

share.plot.area(ax=axes[2], linewidth=0)
axes[2].set_title("Participação mensal das categorias (volume)")
axes[2].set_ylabel("Participação")
axes[2].set_xlabel("")
axes[2].legend(loc="upper left", bbox_to_anchor=(1.01, 1), fontsize=8)

plt.tight_layout()
plt.show()
```

## 6. Top produtos — volume e receita

```python
produto_agg = (
    df.groupby(["recipe_id", "produto", "categoria"], as_index=False)
    .agg(
        vendas=("id", "count"),
        unidades=("quantity", "sum"),
        receita=("receita_linha", "sum"),
        ticket_medio=("receita_linha", "mean"),
    )
)
produto_agg["produto_lbl"] = produto_agg["produto"].str.slice(0, 35)

top_vendas = produto_agg.nlargest(15, "vendas")
top_receita = produto_agg.nlargest(15, "receita")
top_ticket = produto_agg[produto_agg["vendas"] >= 100].nlargest(15, "ticket_medio")
bottom_vendas = produto_agg.nsmallest(15, "vendas")

fig, axes = plt.subplots(2, 2, figsize=(18, 12))

sns.barplot(data=top_vendas, y="produto_lbl", x="vendas", ax=axes[0, 0], palette="Blues_r")
axes[0, 0].set_title("Top 15 por nº de vendas")

sns.barplot(data=top_receita, y="produto_lbl", x="receita", ax=axes[0, 1], palette="Greens_r")
axes[0, 1].set_title("Top 15 por receita (R$)")

sns.barplot(data=top_ticket, y="produto_lbl", x="ticket_medio", ax=axes[1, 0], palette="Purples_r")
axes[1, 0].set_title("Top 15 por ticket médio (vendas ≥ 100)")

sns.barplot(data=bottom_vendas, y="produto_lbl", x="vendas", ax=axes[1, 1], palette="Oranges_r")
axes[1, 1].set_title("15 produtos com menor volume")

for ax in axes.ravel():
    ax.set_xlabel("")
    ax.set_ylabel("")

plt.tight_layout()
plt.show()

top_vendas[["recipe_id", "produto", "categoria", "vendas", "unidades", "receita", "ticket_medio"]].head(10)
```

## 7. Evolução do mix — top 5 produtos vs. demais

```python
top5_ids = df["recipe_id"].value_counts().head(5).index.tolist()
df["top5_grupo"] = np.where(df["recipe_id"].isin(top5_ids), df["recipe_id"], "Outros")

name_map = dict(zip(produtos["recipe_id"], produtos["produto"]))

mix_mensal = (
    df.assign(periodo=df["date_time"].dt.to_period("M").dt.to_timestamp())
    .groupby(["periodo", "top5_grupo"])["id"].count()
    .groupby(level=0).apply(lambda x: x / x.sum() * 100)
    .unstack(fill_value=0)
)
mix_mensal.columns = [
    (name_map.get(c, c)[:30] if c != "Outros" else "Outros") for c in mix_mensal.columns
]

fig, axes = plt.subplots(1, 2, figsize=(20, 6))

mix_mensal.plot.area(ax=axes[0], alpha=0.85, linewidth=0)
axes[0].set_title("Participação mensal — top 5 vs. demais")
axes[0].set_ylabel("% das vendas")
axes[0].set_xlabel("")
axes[0].legend(loc="upper left", bbox_to_anchor=(1.01, 1), fontsize=8)

share10 = df["recipe_id"].value_counts(normalize=True).head(10) * 100
labels10 = [name_map.get(i, i)[:30] for i in share10.index]
axes[1].pie(share10.values, labels=labels10, autopct="%1.1f%%", startangle=140,
             textprops={"fontsize": 9})
axes[1].set_title("Participação dos top 10 produtos")

plt.tight_layout()
plt.show()
```

## 8. Impacto de feriados e eventos

O `resumo_diario_vendas.csv` traz flags por dia. Comparamos a média de vendas diárias com
vs. sem cada flag.

```python
rd = resumo_diario.copy().rename(columns={"vendas_dia": "vendas"})

flags = [
    ("is_holiday", "Feriado"),
    ("is_carnaval_window", "Janela Carnaval"),
    ("is_sao_joao", "São João (20-25/jun)"),
    ("is_summer", "Verão (dez-mar)"),
    ("is_promo_day", "Dia promocional"),
    ("is_rain_event", "Dia de chuva"),
]
flags = [(c, t) for c, t in flags if c in rd.columns]

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
axes = axes.ravel()

for i, (col, titulo) in enumerate(flags):
    stats = rd.groupby(col)["vendas"].mean()
    labels = ["Não", "Sim"] if set(stats.index) == {0, 1} else stats.index.astype(str)
    cores = ["#4C72B0", "#DD8452"][: len(stats)]
    axes[i].bar(labels[: len(stats)], stats.values, color=cores, edgecolor="white")
    axes[i].set_title(f"{titulo}")
    axes[i].set_ylabel("Vendas médias/dia")
    for j, v in enumerate(stats.values):
        axes[i].text(j, v + max(stats.values) * 0.01, f"{v:.0f}", ha="center", fontsize=10)

for j in range(len(flags), len(axes)):
    axes[j].axis("off")

plt.suptitle("Impacto das flags sazonais — média de vendas diárias", y=1.02, fontsize=14)
plt.tight_layout()
plt.show()

# Tabela detalhada por tipo de feriado (merge em 'data' nos dois lados)
diario_feriado = diario.merge(feriados, left_on="data", right_on="data", how="left")
diario_feriado["tipo"] = diario_feriado["tipo"].fillna("DIA NORMAL")

tipo_stats = (
    diario_feriado.groupby("tipo")
    .agg(dias=("vendas", "count"), vendas_medias=("vendas", "mean"))
    .sort_values("vendas_medias", ascending=False)
    .round(1)
)
tipo_stats
```

## 9. Impacto da indisponibilidade de produtos

Os produtos listados em `produtos_indisponiveis.csv` não devem aparecer em vendas durante o
intervalo definido. Validamos isso e mostramos o comportamento antes/depois.

```python
finals = receitas[receitas["type"].str.strip().str.upper() == "PRODUTO_FINAL"].copy()
finals["id"] = finals["id"].astype(str)
finals["name_u"] = finals["name"].astype(str).str.upper()

# Resolve recipe_ids afetados por cada janela de indisponibilidade
janelas = []
for _, row in indisp.iterrows():
    token = str(row["match"]).strip()
    if not token:
        continue
    if token in set(finals["id"]):
        ids = {token}
    else:
        ids = set(finals.loc[finals["name_u"].str.contains(token.upper(), na=False, regex=False), "id"])
    if ids:
        janelas.append({
            "match": token,
            "data_inicio": row["data_inicio"],
            "data_fim": row["data_fim"],
            "recipe_ids": ids,
        })

# Verifica congruência: zero vendas dentro da janela?
rows = []
for jan in janelas:
    sub = df[df["recipe_id"].isin(jan["recipe_ids"])]
    dentro = sub[(sub["data"] >= jan["data_inicio"]) & (sub["data"] <= jan["data_fim"])]
    fora = sub[(sub["data"] < jan["data_inicio"]) | (sub["data"] > jan["data_fim"])]
    dias_dentro = max((jan["data_fim"] - jan["data_inicio"]).days + 1, 1)
    dias_total = (df["data"].max() - df["data"].min()).days + 1
    dias_fora = max(dias_total - dias_dentro, 1)
    rows.append({
        "match": jan["match"],
        "data_inicio": jan["data_inicio"].date(),
        "data_fim": jan["data_fim"].date(),
        "n_receitas": len(jan["recipe_ids"]),
        "vendas_dentro": len(dentro),
        "vendas_fora": len(fora),
        "vendas_dia_fora": round(len(fora) / dias_fora, 2),
        "vendas_dia_dentro": round(len(dentro) / dias_dentro, 2),
    })
indisp_stats = pd.DataFrame(rows).sort_values("vendas_fora", ascending=False)
display(indisp_stats)

# 8 painéis: produto representativo de cada janela com o período sombreado
if janelas:
    n_panels = min(len(janelas), 8)
    fig, axes = plt.subplots((n_panels + 1) // 2, 2, figsize=(16, 3 * ((n_panels + 1) // 2)), sharex=True)
    axes = np.atleast_1d(axes).ravel()
    for i, jan in enumerate(janelas[:n_panels]):
        sub = df[df["recipe_id"].isin(jan["recipe_ids"])]
        if sub.empty:
            axes[i].axis("off")
            continue
        diaria = sub.groupby("data", as_index=False)["id"].count().rename(columns={"id": "vendas"})
        axes[i].plot(diaria["data"], diaria["vendas"].rolling(7, min_periods=1).mean(), linewidth=1.4)
        axes[i].axvspan(jan["data_inicio"], jan["data_fim"], color="red", alpha=0.15)
        first_name = next(iter(jan["recipe_ids"]))
        title = name_map.get(first_name, first_name)[:40]
        axes[i].set_title(f"{jan['match']} | {title}")
        axes[i].grid(alpha=0.3)
    for j in range(n_panels, len(axes)):
        axes[j].axis("off")
    plt.tight_layout()
    plt.show()
```

## 10. Coerência preço x volume

Esperamos uma correlação negativa entre preço e volume: itens mais caros vendem menos.

```python
scatter = produto_agg.copy()
preco_map = dict(zip(produtos["recipe_id"], produtos["preco"]))
scatter["preco"] = scatter["recipe_id"].map(preco_map)
scatter = scatter[(scatter["vendas"] > 0) & (scatter["preco"] > 0)].copy()

fig, ax = plt.subplots(figsize=(11, 6))
sns.scatterplot(
    data=scatter,
    x="preco",
    y="vendas",
    hue="categoria",
    size="receita",
    sizes=(40, 400),
    alpha=0.75,
    ax=ax,
)
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_title("Preço unitário × volume de vendas (escala log)")
ax.set_xlabel("Preço unitário (R$, log)")
ax.set_ylabel("Nº de vendas (log)")
ax.legend(loc="upper left", bbox_to_anchor=(1.01, 1), fontsize=8)
plt.tight_layout()
plt.show()

spearman = scatter["preco"].rank().corr(scatter["vendas"].rank())
pearson_log = np.log(scatter["preco"]).corr(np.log(scatter["vendas"]))
print(f"Correlação Spearman preço×vendas: {spearman:.3f}")
print(f"Correlação Pearson (log×log): {pearson_log:.3f}")
scatter[["recipe_id", "produto", "categoria", "preco", "vendas", "receita"]].sort_values("receita", ascending=False).head(10)
```
