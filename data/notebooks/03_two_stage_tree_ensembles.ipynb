# Notebook 3 - Modelos Baseados em Arvores

Pipeline two-stage com Random Forest como modelo principal e XGBoost/Gradient Boosting como baselines.

```python
from pathlib import Path
import sys

NOTEBOOK_DIR = Path.cwd()
if (NOTEBOOK_DIR / "two_stage_common.py").exists():
    sys.path.insert(0, str(NOTEBOOK_DIR))
else:
    sys.path.insert(0, str((Path("data") / "notebooks").resolve()))

from sklearn.ensemble import (
    GradientBoostingClassifier,
    GradientBoostingRegressor,
    RandomForestClassifier,
    RandomForestRegressor,
)
from xgboost import XGBClassifier, XGBRegressor

from two_stage_common import run_training_notebook
```

```python
classifiers = {
    "Random Forest Classifier": RandomForestClassifier(),
    "XGBoost Classifier": XGBClassifier(),
    "Gradient Boosting Classifier": GradientBoostingClassifier(),
}

regressors = {
    "Random Forest Regressor": RandomForestRegressor(),
    "XGBoost Regressor": XGBRegressor(),
    "Gradient Boosting Regressor": GradientBoostingRegressor(),
}
```

```python
results = run_training_notebook(
    notebook_id="03_tree_ensembles",
    family="Arvores",
    classifiers=classifiers,
    regressors=regressors,
)

display(results["classification_metrics"])
display(results["regression_metrics"])
display(results["summary"][["notebook", "family", "sample_shape", "split_counts", "positive_rate", "feature_count"]])
```
